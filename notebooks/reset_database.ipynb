# Database Reset Utility

This notebook drops all tables from the database and recreates them. It is intended for **testing purposes**, allowing you to reset the database state completely so you can test all repositories without data conflicts.

```python
# Import database configurations and all models.
from domain.database import Base, engine
from sqlalchemy.exc import SQLAlchemyError

# Importing all models ensures that SQLAlchemy registers their metadata before calling drop_all / create_all.
from domain.models import *

def clear_database():
    """
    Deletes all content from the database by dropping all tables and recreating them empty.
    """
    try:
        # Drop all tables currently bound to the Base metadata
        Base.metadata.drop_all(bind=engine)
        
        # Recreate all tables freshly (without any data)
        Base.metadata.create_all(bind=engine)
        
        # Success message
        print("SUCCESS: Database content has been cleared successfully! (All tables dropped and recreated)")
        
    except SQLAlchemyError as e:
        # Catch any database related errors (e.g. constraints, connection issues)
        print(f"ERROR: A database error occurred while trying to clear the DB:\n{e}")
    except Exception as e:
        # Catch any other unexpected errors
        print(f"CRITICAL ERROR: An unexpected error occurred:\n{e}")

# Execute the function to clear the database.
clear_database()
```
